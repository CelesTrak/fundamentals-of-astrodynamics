## Problem Definition
---

GIVEN:&ensp; $i_{initial}$ = 55°, $i_{final}$ = 40°, $\Delta{\Omega}$ = 45°, $v_{initial}$ = 5.892311 km/s (0.745356 ER/TU), $e$ = 0.0 ($\phi_{fpa}$ = 0.0)<br>
FIND: &emsp;$\Delta v_{i{\textit{\ and}} \ \Omega}$

```python
i_init = np.radians(55)   # rad
i_final = np.radians(40)  # rad
draan = np.radians(45)   # rad
v_init = 5.892311        # km/s
fpa = 0                  # flight path angle, rad
```

## Solution
---

**Algorithm 41** summarizes the combined changes to inclination and ascending node for ciruclar orbits only.

First, find the angle between the velocity vectors of the initial and final orbits using spherical trigonometry:

$$
\cos(\vartheta) = \cos(i_{initial}) \cos(i_{final}) + \sin(i_{initial}) \sin(i_{final}) \cos(\Delta{\Omega})
$$

Then, find the change in velocity using **Equation 6-24**:

$$
\Delta v_{i{\textit{\ and}} \ \Omega} = 2 \ v_{initial} \sin\left(\frac{\vartheta}{2}\right)
$$

The location of the burn (argument of latitude) can be determined from **Equation 6-25** and **Equation 6-26**:

$$
\begin{aligned}
\cos(u_{initial}) &= \frac
{\sin(i_{final})\cos(\Delta{\Omega}) - \cos(\vartheta)\sin(i_{initial})}
{\sin(\vartheta)\cos(i_{initial})} \\
\\
\cos(u_{final}) &= \frac
{\cos(i_{initial})\sin(i_{final}) - \sin(i_{initial})\cos(i_{final})\cos(\Delta{\Omega})}
{\sin(\vartheta)}
\end{aligned}
$$

This can all be done with the `incl_and_node` routine:

```python
dv_incl_node, arglat_init, arglat_final = incl_and_node(i_init, i_final, draan, v_init, fpa)

print(f'delta-v (incl + node):\t\t{dv_incl_node:.6f}\tkm/s\n')
print(f'Arg. of Latitude (initial):\t{np.degrees(arglat_init):.4f}\tdeg')
print(f'Arg. of Latitude (final):\t{np.degrees(arglat_final):.4f}\t\tdeg')
```
