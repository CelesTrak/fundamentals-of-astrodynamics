## Problem Definition
---

GIVEN:&ensp; Initial orbit &emsp;alt = 191.34411 km<br>
&ensp;&ensp;&emsp;&emsp;&emsp;Final orbit &emsp;  alt = 35,781.34857 km<br>
FIND: &emsp;$\Delta{v}$, time of flight ($\Delta{t}$)

```python
alt_init = 191.34411
alt_final = 35781.34857
e_init = e_final = 0
nu_init = 0
nu_final = np.pi
```

## Solution
---

Begin by determining radii values:

```python
r_init = const.RE + alt_init
r_final = const.RE + alt_final

print(f'r_initial:\t{r_init}\tkm')
print(f'r_final:\t{r_final}\tkm')
```

**Algorithm 36** summarizes the Hohmann transfer process.

First, find the velocities at each position in the transfer. From **Equation 1-33** for a satellite's circular speed:

$$
\begin{aligned}
v_{initial} &= \sqrt{\frac{\mu}{r_{initial}}} \\
\\
v_{final} &= \sqrt{\frac{\mu}{r_{final}}}
\end{aligned}
$$

For the transfer orbit:

$$
a_{trans} = \frac{r_{initial} + r_{final}}{2}
$$

From **Equation 1-31** for elliptical satellite speed, find the two transfer velocities:

$$
\begin{aligned}
v_{trans_{a}} &= \sqrt{\frac{2\mu}{r_{initial}} - \frac{\mu}{a_{trans}}} \\
\\
v_{trans_{b}} &= \sqrt{\frac{2\mu}{r_{final}} - \frac{\mu}{a_{trans}}} 
\end{aligned}
$$

The total change in velocity is the sum of the two burns. Notice that each velocity is positive, indicating increasing semimajor axis values.

$$
\Delta{v} = \Delta{v_a} + \Delta{v_b} = \left| v_{trans_{a}} - v_{initial} \right| + \left| v_{final} - v_{trans_{b}} \right|
$$

Finally, the time of flight for a Hohmann transfer is half the period of the transfer orbit:

$$
\tau_{trans} = \pi \ \sqrt{\frac{a_{trans}^3}{\mu}}
$$

This can all be achieved with the `hohmann` routine:

```python
delta_v_a, delta_v_b, dt_sec = hohmann(r_init, r_final, e_init, e_final, nu_init, nu_final)

print(f'delta-v (a):\t{delta_v_a:.4f}\tkm/s')
print(f'delta-v (b):\t{delta_v_b:.4f}\tkm/s')
print(f'delta-v (tot):\t{delta_v_a + delta_v_b:.4f}\tkm/s\n')
print(f'Time of flight:\t{dt_sec/const.MIN2SEC:.3f}\tmin')
```
