## Problem Definition
---

GIVEN:&ensp; $i$ = 55.0°, $\Delta{\Omega}$ = 45.0°, $v_{initial}$ = 5.892311 km/s (0.745356 ER/TU), $e_{initial}$ = 0.0 ($\phi_{fpa}$ = 0.0)<br>
FIND: &emsp;$\Delta v_{\Omega{\textit{\ only}}}$

```python
i_init = np.radians(55)  # rad
draan = np.radians(45)   # rad
v_init = 5.892311        # km/s
ecc = 0
fpa = 0                  # flight path angle, rad
```

## Solution
---

**Algorithm 40** summarizes the change in the ascending node for ciruclar orbits only.

First, determine the angle through which the velocity is rotated by using **Equation 6-23**:

$$
\cos(\vartheta) = \cos^2(i_{initial}) + \sin^2(i_{initial}) \cos(\Delta{\Omega})
$$

Then, find the change in velocity using **Equation 6-24**:

$$
\Delta{v_{\Omega{\textit{\ only}_{cir}}}} = 2 \ v_{initial} \sin\left(\frac{\vartheta}{2}\right)
$$

The argument of latitude for the common points can be determined from **Equation 6-21** and **Equation 6-22**:

$$
\begin{aligned}
\cos(u_{initial}) &= \tan(i_{initial}) \frac{\cos(\Delta\Omega) - \cos(\vartheta)}{\sin(\vartheta)} \\
\\
\cos(u_{final}) &= \cos(i_{initial}) \sin(i_{initial}) \left( \frac{1 - \cos(\Delta\Omega)}{\sin(\vartheta)} \right)
\end{aligned}
$$

This can all be done with the `node_only` routine:

```python
_, dv_node_only, arglat_init, arglat_final = node_only(i_init, ecc, draan, v_init, fpa, incl=i_init)

print(f'delta-v (node only):\t\t{dv_node_only:.6f}\tkm/s\n')
print(f'Arg. of Latitude (initial):\t{np.degrees(arglat_init):.4f}\tdeg')
print(f'Arg. of Latitude (final):\t{np.degrees(arglat_final):.4f}\t\tdeg')
```
