## Problem Definition
---

GIVEN:&ensp; $a_{int} = a_{tgt}$ = 12,756.2726 km, $\vartheta$ = -20° (target in front of interceptor)<br>
FIND: &emsp;$\Delta{v}$

```python
a_interceptor = a_target = 12756.2726
phase_angle = np.radians(-20)
e_init = e_final = 0  # circular orbit
nu_init = nu_final = 0
```

## Solution
---

**Algorithm 44** summarizes the techniques for determining the circular coplanar phasing for two satellites in the same orbit.

Because the satellitse are in the same orbit, the angular velocity of the interceptor orbit ($\omega_{int}$) is the same as target's angular velocity ($\omega_{tgt}$), which you should recognize as the mean motion, _n_:

$$
\omega_{tgt} = \sqrt{\frac{\mu}{a_{tgt}^3}} = n
$$

The phasing time formula is then:

$$
\tau_{phase} = \Delta{t} = t_1 - t_0 = \frac{k_{tgt} (2\pi) + \vartheta}{\omega_{tgt}}
$$

where $k_{tgt}$ signifies the number of target revolutions.

Because the phasing orbit consumes one of the interceptor's satellite's periods as it accomplishes the rendezvous, we may solve the time-period equation (**Equation 1-26**) for the semimajor axis:

$$
a_{phase} = \left(
\mu
\left(
\frac{\tau_{phase}}{k_{int}(2\pi)}
\right)^2
\right)^{1/3}
$$

Now determine the change in velocity for the maneuver. Because you're examining only circular orbits with tangential burns, the change in velocity is simply the change in magnitudes of the different orbits.

For the first burn:

$$
\Delta{v} = v_{phase} - v_{int}
$$

For the second burn:

$$
\Delta{v} = v_{int} - v_{phase}
$$

Notice both changes in velocities have the same magnitudes but different directions.

Thus, the total $\Delta{v}$ is twice the change in velocity of the initial burn:

$$
\Delta{v} = 2 \left| v_{phase} - v_{int} \right|
= 2 \left|
\sqrt{\frac{2 \mu}{a_{tgt}} - \frac{\mu}{a_{phase}}} -
\sqrt{\frac{\mu}{a_{tgt}}}
\right|
$$

We can accomplish this with the `rendezvous_coplanar` method routine for a given number of interceptor and target orbits to wait.

For the case where both the interceptor and target complete one revolution ($k_{int} = k_{tgt} = 1$):

```python
k_int = k_tgt = 1

_, t_phase, delta_v = rendezvous_coplanar(
    a_interceptor, a_target, phase_angle, e_init, e_final, nu_init, nu_final, k_int, k_tgt
)

print(f'τ_phase (wait time):\t{t_phase/const.MIN2SEC:.4f}\tmin')
print(f'Total Δv (rendezvous):\t{delta_v:.6f}\tkm/s')
```

Note the negative sign indicates firing opposite the velocity vector.

Now evaluate the solution for the case in which the interceptor and target complete 2 revolutions ($k_{int} = k_{tgt} = 2$):

```python
k_int = k_tgt = 2

_, t_phase, delta_v = rendezvous_coplanar(
    a_interceptor, a_target, phase_angle, e_init, e_final, nu_init, nu_final, k_int, k_tgt
)

print(f'τ_phase (wait time):\t{t_phase/const.MIN2SEC:.4f}\tmin')
print(f'Total Δv (rendezvous):\t{delta_v:.6f}\tkm/s')
```

Notice the large reduction in velocity when you use two phasing orbits!
