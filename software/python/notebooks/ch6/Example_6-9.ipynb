## Problem Definition
---

GIVEN:&ensp; $a_{int}$ = 12,756.2726 km, $a_{tgt}$ = 42,159.48 km, $\vartheta$ = -20° (target in front of interceptor)<br>
FIND: &emsp;$\Delta{v}$

```python
a_interceptor = 12756.2726     # km
a_target = 42159.48            # km
phase_angle = np.radians(-20)  # rad
e_init = e_final = 0           # circular orbit
nu_init = nu_final = 0         # rad
```

## Solution
---

**Algorithm 45** summarizes the techniques for determining the circular coplanar phasing for two satellites in different orbits.

Find the angular velocities (mean motions) of the two satellites, using the radii as the semimajor axis because the orbits are circular:

$$
\begin{aligned}
\omega_{tgt} &= \sqrt{\frac{\mu}{a_{tgt}^3}} \\
\omega_{int} &= \sqrt{\frac{\mu}{a_{int}^3}} \\
\end{aligned}
$$

The semimajor axis of the phasing orbit is:

$$
a_{trans} = \frac{a_{int} + a_{tgt}}{2}
$$

The time required for the transfer is half the period of the Hohmann transfer, or:

$$
\tau_{trans} = \pi \ \sqrt{\frac{a_{trans}^3}{\mu}}
$$


The lead angle the target satellite must have for the transfer to take place is:

$$
\alpha_L = \omega_{tgt} \ \tau_{trans}
$$

The phasing angle for the target satellite must have for the transfer to take place is:

$$
\vartheta = \pi - \alpha_L
$$

Remember that this must be positive!

The wait time, or the time until the interceptor and target are in the correct positions, depends on the phase-angle separations (initial and final) and the relative velocities of the two satellites:

$$
\tau_{wait} = \frac{\vartheta - \vartheta_i + 2 \pi k}{\omega_{int} - \omega_{tgt}}
$$

where $\vartheta_i$ is the initial phase angle, $\omega_{int}$ is the interceptor's angular velocity, and $\frac{2\pi}{\omega_{int} - \omega_{tgt}}$ is the time during which the relative geometry repeats (called the _synodic period_). _k_ denotes the number of revolutions.

Finally, determine the total change in velocity using the equations for the Hohmann's transfer:

$$
\Delta{v} = \left|
\sqrt{\frac{2 \mu}{a_{int}} - \frac{\mu}{a_{trans}}} -
\sqrt{\frac{\mu}{a_{int}}}
\right| +
\left|
\sqrt{\frac{2 \mu}{a_{tgt}} - \frac{\mu}{a_{trans}}} -
\sqrt{\frac{\mu}{a_{tgt}}}
\right|
$$

We can accomplish this with the `rendezvous_coplanar` method routine for a given number of interceptor and target orbits to wait.

For the case where both the interceptor and target complete one revolution ($k_{int} = k_{tgt} = 1$):

```python
k_int = k_tgt = 1

_, t_phase, delta_v = rendezvous_coplanar(
    a_interceptor, a_target, phase_angle, e_init, e_final, nu_init, nu_final, k_int, k_tgt
)

print(f'τ_phase (wait time):\t{t_phase/const.MIN2SEC:.4f}\tmin')
print(f'Total Δv (rendezvous):\t{delta_v:.6f}\tkm/s')
```
