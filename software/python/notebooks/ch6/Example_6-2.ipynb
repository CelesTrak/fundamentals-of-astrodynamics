## Problem Definition
---

GIVEN:&ensp; Initial orbit &emsp;alt = 191.34411 km<br>
&ensp;&ensp;&emsp;&emsp;&emsp;Point _b_ &emsp;&emsp;&ensp; alt = 503,873 km (79 ER)<br>
&ensp;&ensp;&emsp;&emsp;&emsp;Final orbit &emsp;&ensp;alt = 376,310 km (59 ER)<br>
FIND: &emsp;$\Delta{v}$, time of flight ($\Delta{t}$)

```python
alt_init = 191.34411  # km
alt_b = 503873        # km
alt_final = 376310    # km
e_init = e_final = 0
nu_init = 0           # rad
nu_final = np.pi      # rad
```

## Solution
---

Begin by determining radii values:

```python
r_init = const.RE + alt_init
r_b = const.RE + alt_b
r_final = const.RE + alt_final

print(f'r_initial:\t{r_init}\tkm\t({r_init/const.RE:.2f} ER)')
print(f'r_b:\t\t{r_b}\tkm\t({r_b/const.RE:.2f} ER)')
print(f'r_final:\t{r_final}\tkm\t({r_final/const.RE:.2f} ER)')
```

**Algorithm 37** summarizes the bi-elliptic transfer process.

First, find the velocities at each position in the transfer. From **Equation 1-33** for a satellite's circular speed:

$$
\begin{aligned}
v_{initial} &= \sqrt{\frac{\mu}{r_{initial}}} \\
\\
v_{final} &= \sqrt{\frac{\mu}{r_{final}}}
\end{aligned}
$$

For the transfer orbits:

$$
\begin{aligned}
a_{trans1} = \frac{r_{initial} + r_b}{2} \\
\\
a_{trans2} = \frac{r_b + r_{final}}{2}
\end{aligned}
$$

From **Equation 1-31** for elliptical satellite speed, find the transfer velocities:

$$
\begin{aligned}
v_{trans1_{a}} &= \sqrt{\frac{2\mu}{r_{initial}} - \frac{\mu}{a_{trans1}}} \\
\\
v_{trans1_{b}} &= \sqrt{\frac{2\mu}{r_b} - \frac{\mu}{a_{trans1}}} \\
\\
v_{trans2_{b}} &= \sqrt{\frac{2\mu}{r_b} - \frac{\mu}{a_{trans2}}} \\
\\
v_{trans2_{c}} &= \sqrt{\frac{2\mu}{r_{final}} - \frac{\mu}{a_{trans2}}} \\
\end{aligned}
$$

The total change in velocity is the sum of the three tangential burns. Notice here that the final burn is negative, indicating the satellite must slow down:

$$
\Delta{v} = \Delta{v_a} + \Delta{v_b} + \Delta{v_c} = 
\left| v_{trans1_{a}} - v_{initial} \right| + 
\left| v_{trans2_{b}} - v_{trans1_{b}} \right| + 
\left| v_{final} - v_{trans2_{c}} \right|
$$

Finally, the time of flight for a bi-elliptic transfer is half the period of both the transfer orbits:

$$
\tau_{trans} = \pi \ \sqrt{\frac{a_{trans1}^3}{\mu}} + \pi \ \sqrt{\frac{a_{trans2}^3}{\mu}}
$$

This can all be achieved with the `bielliptic` routine:

```python
delta_v_a, delta_v_b, delta_v_c, dt_sec = bielliptic(r_init, r_b, r_final, e_init, e_final, nu_init, nu_final)

print(f'delta-v (a):\t{delta_v_a:.4f}\tkm/s')
print(f'delta-v (b):\t{delta_v_b:.4f}\tkm/s')
print(f'delta-v (c):\t{delta_v_c:.4f}\tkm/s')
print(f'delta-v (tot):\t{delta_v_a + delta_v_b + delta_v_c:.4f}\tkm/s\n')
print(f'Time of flight:\t{dt_sec/const.HR2SEC:.3f}\thr')
```
